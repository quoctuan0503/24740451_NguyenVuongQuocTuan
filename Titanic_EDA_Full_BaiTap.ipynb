## Câu hỏi 1: Hạng vé nào có tỷ lệ sống sót cao nhất, chênh lệch bao nhiêu so với hạng thấp nhất?

````python
pclass_survival = df.groupby('pclass')['survived'].mean()
diff = pclass_survival.max() - pclass_survival.min()

print(pclass_survival)
print(f"\nCao nhất: Hạng {pclass_survival.idxmax()} ({pclass_survival.max():.2%})")
print(f"Chênh lệch: {diff:.2%}")
````

## Câu hỏi 2: Giới tính hay hạng vé ảnh hưởng đến sống sót mạnh hơn?

````python
sex_diff = df.groupby('sex')['survived'].mean().pipe(lambda s: s.max() - s.min())
pclass_diff = df.groupby('pclass')['survived'].mean().pipe(lambda s: s.max() - s.min())

print("Chênh lệch theo Giới tính (Sex):", sex_diff)
print("Chênh lệch theo Hạng vé (Pclass):", pclass_diff)
````
